# On LUMI

\# Build container using cotainr 

module load cotainr

cotainr build gpu_ocean.sif --system=lumi-g --conda-env=conda_environment.yml

sbatch job/start_ipyparallel.job \# start ipyparallel's ipcontroller and ipengines

salloc -p dev-g --account=project_465002280 --gpus-per-node=2 --ntasks-per-node=2 --cpus-per-task=2 --mem-per-gpu=20G --time=02:00:00 \# allocation to run jupyter lab in

squeue -u # get \<hostname\>

srun singularity exec -B /project/project_465002280/martinli/gpuocean:/notebooks /project/project_465002280/martinli/gpuocean/gpu_ocean.sif jupyter lab --allow-root --ip 0.0.0.0 --notebook-dir /notebooks --NotebookApp.token='GPUOcean!1' \# start jupyter lab
