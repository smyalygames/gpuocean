```python
import cupy as cp
from mpi4py import MPI
import numpy as np
```

```python


comm = MPI.COMM_WORLD
rank = comm.Get_rank()
size = comm.Get_size()

if size < 2:
    if rank == 0:
        print("This test requires at least 2 MPI processes.")
    exit()

# Allocate a CuPy array on the default device (GPU)
# Make sure the buffer is ready before MPI operations using a synchronization stream
n = 100
gpu_array = cp.arange(n, dtype=cp.float32)
cp.cuda.Stream.null.synchronize() # Synchronize the stream

if rank == 0:
    print(f"Rank {rank}: Sending array from GPU memory to Rank 1")
    # Send the CuPy array directly
    comm.Send([gpu_array, MPI.FLOAT], dest=1, tag=77)
    print(f"Rank {rank}: Sent data.")

elif rank == 1:
    # Create an empty CuPy array to receive data
    recv_array = cp.empty(n, dtype=cp.float32)
    print(f"Rank {rank}: Receiving array into GPU memory from Rank 0")
    # Receive directly into the CuPy array
    comm.Recv([recv_array, MPI.FLOAT], source=0, tag=77)
    cp.cuda.Stream.null.synchronize() # Synchronize the stream
    print(f"Rank {rank}: Received data. Data check: {cp.allclose(recv_array, cp.arange(n, dtype=cp.float32))}")
```
