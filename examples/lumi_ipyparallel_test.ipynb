```python
import ipyparallel as ipp
rc = ipp.Client()
view = rc[:]
view.block = True

# Get the MPI rank from each engine
ranks = view.apply_sync(lambda: __import__('mpi4py.MPI').MPI.COMM_WORLD.Get_rank())
print(ranks) 
# Expected output is a list of unique ranks, e.g., [0, 1, ...]
```
